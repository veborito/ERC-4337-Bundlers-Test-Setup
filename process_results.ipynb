```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
def read_data_cpu(file_path):
  data = pd.read_csv(file_path)
  return data[data["scope"] == "cpu"]
```

```python
anvil_data_cpu = read_data_cpu("./retest_xeon_gold/AnvilTest8PowerReport.csv")
alto_data_cpu = read_data_cpu("./retest_xeon_gold/AltoTest8PowerReport.csv")
alto_avg_power = alto_data_cpu["power"].mean()
anvil_avg_power = anvil_data_cpu["power"].mean()
print(f"Alto average power test 5: {alto_avg_power}")
print(f"Anvil average power test 5: {anvil_avg_power}")
```
